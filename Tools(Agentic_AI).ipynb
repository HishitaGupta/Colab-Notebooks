# Custom Tools (Method 1)

```python
from langchain_core.tools import tool
```

```python
# Step 1 -- create a function

def multiply(a,b):
  return a*b
```

```python
# Step 2 -- Adding type hints

def multiply(a:int,b:int)->int:
  return a*b
```

```python
# Step 3 -- Adding tooldecorator

@tool("multiply_numbers", return_direct=True)
def multiply_numbers(a:int,b:int)->int:
  """
    Multiplies two numbers and returns the result.
    Args:
        a (int): First number
        b (int): Second number
    Returns:
        int: Product of a and b
    """
  return a*b
```

```python
multiply_numbers.invoke({"a": 5, "b": 6})
```

```python
# LLM expects tool to respond back in json format

print(multiply_numbers.args_schema.model_json_schema())
```

# Structured Tool with the help of Pydantic

```python
from langchain_core.tools import StructuredTool
# from pydantic import BaseModel, Field
```

```python
# Creating a pydantic class
class MultiplyInput(BaseModel):
  a: int = Field(required=True, description = "The first Number to add")
  b: int = Field(required=True, description = "The second Number to add")
```

```python
def multiply_func(a:int, b:int) -> int:
  return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func = multiply_func,
    name="multiply",
    description ="Multiply two numbers",
    args_schema = MultiplyInput

)
```

```python
result = multiply_tool.invoke({'a':9, 'b':10})
```

```python
result
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

# Custom Tools using BaseTools

```python
from langchain_core.tools import BaseTool
```

```python
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
  a: int = Field(required=True, description = "The first Number to add")
  b: int = Field(required=True, description = "The second Number to add")
```

```python
class MultiplyTool(BaseTool):
  name:str ="Multiply"
  description:str ="Multiply two numbers"

  args_schema :Type[BaseModel] = MultiplyInput

  def _run(self,a:int,b:int)->int:
    return a*b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':3,'b':3})

result
```

# ToolKit

```python
# Custom tool

@tool
def addition(a:int,b:int)->int:
  """Add Two Numbers"""
  return a+b

@tool
def multiply(a:int,b:int)->int:
  """Multiply Two Numbers"""
  return a*b

@tool
def divide(a:int,b:int)->int:
  """Divide Two Numbers"""
  return a/b
```

```python
class MathToolKit:
  def get_tools(self):
    return[addition,multiply,divide]
```

```python
toolkit = MathToolKit()
tools = toolkit.get_tools()

for tool in tools:
  print(tool.name)
```
